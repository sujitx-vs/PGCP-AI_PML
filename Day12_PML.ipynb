First, let's prepare the data by scaling the numerical features and applying PCA for dimensionality reduction, which will be useful for visualizing the clusters.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Select numerical columns for clustering (excluding 'city_name' and 'country' since they are label encoded)
numerical_cols = ['population_density', 'avg_income', 'internet_penetration', 'avg_rent', 'air_quality_index', 'public_transport_score', 'happiness_score', 'green_space_ratio']
df_scaled = df[numerical_cols].copy()

# Scale the numerical features
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_scaled)

# Apply PCA for dimensionality reduction for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(df_scaled)

display(X_pca[:5])
```

Now, let's apply K-Means clustering and add the cluster labels to the DataFrame.

```python
# Fit K-Means on the scaled data and add cluster labels to df
k_mean = KMeans(n_clusters=5, random_state=42, n_init=10) # Added random_state and n_init for reproducibility
df['Kmeans_cluster'] = k_mean.fit_predict(df_scaled)
display(df.head())
```

Next, let's apply DBSCAN clustering and add its cluster labels to the DataFrame.

```python
# Apply DBSCAN clustering
dbscan = DBSCAN(eps=0.8, min_samples=5)
df['DBSCAN_cluster'] = dbscan.fit_predict(df_scaled)
display(df.head())
```

Finally, here is the corrected plotting code to visualize the K-Means and DBSCAN clusters using the PCA-reduced data.

```python
plt.figure(figsize=(15,5))

plt.subplot(1,2,1)
for cluster_id in sorted(df['Kmeans_cluster'].unique()):
  mask_k = df['Kmeans_cluster'] == cluster_id
  plt.scatter(X_pca[mask_k, 0], X_pca[mask_k, 1], label=f"Cluster {cluster_id}", s=25)
plt.title('K-Means Clusters (PCA-reduced)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()

plt.subplot(1,2,2)
for cluster_id in sorted(df['DBSCAN_cluster'].unique()):
  mask_d = df['DBSCAN_cluster'] == cluster_id
  if cluster_id == -1:
    label = 'Noise'
    color = 'black'
  else:
    label = f'Cluster {cluster_id}'
    color = None # matplotlib will cycle colors
  plt.scatter(X_pca[mask_d, 0], X_pca[mask_d, 1], label=label, s=25, color=color)
plt.title('DBSCAN Clusters (PCA-reduced)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

# Filter out noise points (label -1) for Silhouette Score calculation
non_noise_indices = df['DBSCAN_cluster'] != -1
X_pca_filtered = X_pca[non_noise_indices]
dbscan_labels_filtered = df['DBSCAN_cluster'][non_noise_indices]

# Calculate Silhouette Score if there are more than one cluster (excluding noise)
if len(set(dbscan_labels_filtered)) > 1:
    silhouette_avg_dbscan = silhouette_score(X_pca_filtered, dbscan_labels_filtered)
    print(f"Silhouette Score for DBSCAN: {silhouette_avg_dbscan}")
else:
    print("Cannot calculate Silhouette Score for DBSCAN: Less than 2 clusters found (excluding noise points).")
```

### Elbow Method to find optimal K

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

sse = []
k_range = range(1, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(inp)
    sse.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(k_range, sse, marker='o')
plt.title('Elbow Method For Optimal K')
plt.xlabel('Number of clusters (K)')
plt.ylabel('Sum of Squared Errors (SSE)')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```

### Silhouette Method to find optimal K

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []
k_range_silhouette = range(2, 11) # Silhouette Score requires at least 2 clusters

for k in k_range_silhouette:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(inp)
    score = silhouette_score(inp, cluster_labels)
    silhouette_scores.append(score)

plt.figure(figsize=(10, 6))
plt.plot(k_range_silhouette, silhouette_scores, marker='o')
plt.title('Silhouette Method For Optimal K')
plt.xlabel('Number of clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(k_range_silhouette)
plt.grid(True)
plt.show()
```

### Determining optimal `eps` for DBSCAN using k-distance graph

For DBSCAN, choosing the right `eps` and `min_samples` is crucial. A common approach to determine `eps` is by plotting the k-distance graph. For each point in the dataset, we calculate the distance to its k-th nearest neighbor. The `k` value is typically set to `min_samples`. In this case, we'll use `min_samples = 2 * n_features` or `n_features + 1`, which for 2 features (`Annual Income`, `Spending Score`) would be around 4 or 5. Let's use `k=4` (distance to the 4th nearest neighbor) to calculate the `eps`.

We plot these k-distances in ascending order, and the 'elbow' or 'knee' in the graph often indicates a good value for `eps`.

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

# Assuming 'inp' is your DataFrame with 'Annual Income (k$)' and 'Spending Score (1-100)'
# For DBSCAN, it's often beneficial to scale the data, though for visual representation,
# we can sometimes get away without it if scales are similar.
# Let's use the 'inp' as is for now, as both features are in similar numerical ranges.

n_neighbors = 4 # A common choice is 2 * n_dimensions, here 2*2=4

neighbors = NearestNeighbors(n_neighbors=n_neighbors)
neighbors_fit = neighbors.fit(inp)
distances, indices = neighbors_fit.kneighbors(inp)

# Sort distances and plot
distances = np.sort(distances[:,n_neighbors-1], axis=0)

plt.figure(figsize=(10, 6))
plt.plot(distances)
plt.title(f'K-distance Graph (k={n_neighbors})')
plt.xlabel('Points sorted by distance to {n_neighbors}th nearest neighbor')
plt.ylabel(f'Distance to {n_neighbors}th Nearest Neighbor')
plt.grid(True)
plt.show()
```

Looking at the k-distance graph (generated in the previous cell), we can identify an 'elbow' point where the slope of the curve changes significantly. This point suggests an appropriate value for `eps`.

To find the `eps` value from the graph:
1.  Locate the point on the curve where there is a sharp bend, resembling an 'elbow' or 'knee'.
2.  Read the corresponding value on the **Y-axis** at this 'elbow'.

From our k-distance graph, this 'elbow' appears to be around **10-12** on the Y-axis. Therefore, we chose `eps=10`.

For `min_samples`, a common choice is `2 * n_features`. In our case, with 2 features (`Annual Income`, `Spending Score`), `2 * 2 = 4`. We opted to use `min_samples=5` which is a good general starting point for 2D data and was also used effectively in our earlier analysis of the City Lifestyle dataset.

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

# Apply DBSCAN clustering with chosen eps and min_samples
dbscan_final = DBSCAN(eps=10, min_samples=5)
mall_data['DBSCAN_Cluster'] = dbscan_final.fit_predict(inp)

# Visualize the clusters
plt.figure(figsize=(10, 8))
sns.scatterplot(data=mall_data, x='Annual Income (k$)', y='Spending Score (1-100)', hue='DBSCAN_Cluster', palette='viridis', s=100)
plt.title('Customer Clusters (DBSCAN)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(title='DBSCAN_Cluster')
plt.grid(True)
plt.show()

display(mall_data.head())
```

### Calculate Silhouette Score for DBSCAN

Similar to K-Means, we can evaluate the quality of the DBSCAN clustering using the Silhouette Score. It's important to exclude noise points (which are labeled as -1 by DBSCAN) when calculating this score, as they do not belong to any cluster.

```python
# Filter out noise points (label -1) for Silhouette Score calculation
non_noise_indices_dbscan = mall_data['DBSCAN_Cluster'] != -1

# Check if there are any non-noise points and more than one cluster
if len(set(mall_data['DBSCAN_Cluster'][non_noise_indices_dbscan])) > 1:
    silhouette_avg_dbscan_mall = silhouette_score(inp[non_noise_indices_dbscan], mall_data['DBSCAN_Cluster'][non_noise_indices_dbscan])
    print(f"Silhouette Score for DBSCAN on Mall Customers: {silhouette_avg_dbscan_mall}")
else:
    print("Cannot calculate Silhouette Score for DBSCAN: Less than 2 clusters found (excluding noise points) or no non-noise points.")
```

### Calculating Silhouette Score for different `eps` values

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Redefine inp for self-contained execution
mall_data = pd.read_csv('/content/Mall_Customers.csv')
inp = mall_data.drop(['CustomerID','Genre','Age'],axis=1)

silhouette_scores_eps = []
eps_range = np.arange(5, 20, 0.5)  # Test eps values from 5 to 20 with a step of 0.5
min_samples_fixed = 5 # Keep min_samples fixed

best_silhouette_score = -1
best_eps = None

for eps_val in eps_range:
    dbscan_temp = DBSCAN(eps=eps_val, min_samples=min_samples_fixed)
    cluster_labels = dbscan_temp.fit_predict(inp)

    # Filter out noise points for Silhouette Score calculation
    non_noise_indices = cluster_labels != -1

    # Check if there are any non-noise points and more than one cluster
    if len(set(cluster_labels[non_noise_indices])) > 1:
        score = silhouette_score(inp[non_noise_indices], cluster_labels[non_noise_indices])
        silhouette_scores_eps.append(score)
        if score > best_silhouette_score:
            best_silhouette_score = score
            best_eps = eps_val
    else:
        silhouette_scores_eps.append(-1) # Append -1 for cases where score cannot be calculated

plt.figure(figsize=(12, 6))
plt.plot(eps_range, silhouette_scores_eps, marker='o')
plt.title('DBSCAN Silhouette Score vs. Epsilon (eps) Value')
plt.xlabel('Epsilon (eps) Value')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.xticks(np.arange(5, 20.5, 1))
plt.show()

print(f"Best Silhouette Score: {best_silhouette_score:.4f} with eps = {best_eps}")
```
