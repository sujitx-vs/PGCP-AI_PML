# Isolation Forest Implementation for one class

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import IsolationForest
```

```python
rng = np.random.RandomState(42)
X_normal = 0.3*rng.randn(200,2)
```

```python
X_outliers = rng.uniform(low=4,high=4,size=(20,2))
```

```python
X = np.vstack([X_normal,X_outliers])
```

```python
model = IsolationForest(contamination=0.2,random_state=42)
model.fit(X)
```

```python
y_pred = model.predict(X)
y_pred
```

```python
normal_points = X[y_pred == 1]
outlier_points = X[y_pred == -1]
```

```python
plt.scatter(normal_points[:,0],normal_points[:,1],color='green')
plt.scatter(outlier_points[:,0],outlier_points[:,1],color='red')
```

```python
import numpy as np
from pyod.models.knn import KNN
```

```python
np.random.seed(42)
X_normal = np.random.randn(300,2)
```

```python
point_anomaly = np.array([[8,8]])
```

```python
X = np.vstack([X_normal,point_anomaly])
```

```python
plt.scatter(X[:,0],X[:,1],color='blue',label="Data Points")
plt.scatter(point_anomaly[:,0],point_anomaly[:,1],color='red',label="Point Anomaly")
plt.legend()
plt.show()
```

```python
clf = KNN(contamination=0.01) #1% of anomaly
clf.fit(X)
```

```python
y_pred = clf.predict(X)
y_pred # 0 is normal and 1 is anomaly
```

```python
score = clf.decision_scores_
score
```

```python
print("Total points :",len(X))
print("Total normal points :",sum(y_pred == 0))
print("Total anomaly points :",sum(y_pred == 1))
```

```python
plt.scatter(X[y_pred==0][:,0],X[y_pred==0][:,1],color="blue",label="Normal Points")
plt.scatter(X[y_pred==1][:,0],X[y_pred==1][:,1],color="red",label="Anomaly Points")
plt.legend()
```

# Collective Anomaly

```python
import numpy as np

from pyod.models.knn import KNN
```

```python
np.random.seed(42)
X_normal = np.random.randn(300,2)
```

```python
coll_anomaly = np.random.uniform(low=-6,high=6,size=(20,2))
```

```python
X = np.vstack([X_normal,coll_anomaly])
```

```python
y_true = np.zeros(len(X),dtype=int)
y_true[-20:] = 1 # last 20 are nominal
```

```python
plt.scatter(X[:,0],X[:,1],color='blue',label="Data Points")
plt.scatter(coll_anomaly[:,0],coll_anomaly[:,1],color='red',label="Collective Anomaly")
plt.legend()
plt.show()
```

```python
clf = KNN(contamination=0.1)
clf.fit(X)
```

```python
y_pred = clf.predict(X)
y_pred
```

```python
score = clf.decision_scores_
score
```

```python
print("Total points :",len(X))
print("Total normal points :",sum(y_pred == 0))
print("Total anomaly points :",sum(y_pred == 1))
```

```python
plt.scatter(X[y_pred==0][:,0],X[y_pred==0][:,1],color="blue",label="Normal Points")
plt.scatter(X[y_pred==1][:,0],X[y_pred==1][:,1],color="red",label="Anomaly Points")
plt.legend()
plt.title("Collective Anomaly Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
from sklearn.metrics import confusion_matrix,classification_report, accuracy_score,precision_score,recall_score,f1_score
print(confusion_matrix(y_true,y_pred))
print("Accuracy :",accuracy_score(y_true,y_pred))
print("Precision :",precision_score(y_true,y_pred))
print("Recall :",recall_score(y_true,y_pred))
print("F1 Score :",f1_score(y_true,y_pred))

print(classification_report(y_true,y_pred))
```

# Contextual Anomaly

```python
import numpy as np
from pyod.models.knn import KNN
```

```python
np.random.seed(42)

tim = np.arange(0,100,0.1)
```

```python
contextual_normal = np.sin(tim)
```

```python
contextual_anomaly = np.sin(tim)
contextual_anomaly[500:600]  = 3 # abnormal spike
```

```python
plt.plot(tim,contextual_normal,label = "Contextual Normal")
plt.plot(tim,contextual_anomaly,label = "Anomaly",linestyle='--')
plt.xlabel("Time")
plt.ylabel("Value")
plt.legend()
plt.show()
```

```python
mean = np.mean(contextual_anomaly)
std =np.std(contextual_anomaly)
```

```python
z_score = (contextual_anomaly - mean)/std
# z_score
```

```python
threshold = 3
anomalies = np.abs(z_score)>threshold
```

```python
plt.plot(tim,contextual_anomaly,label="Signal")
plt.scatter(tim[anomalies],contextual_anomaly[anomalies],color='red',label="Contextual Anomaly")
plt.xlabel("Time")
plt.ylabel("Value")
plt.legend()
plt.show()
```

```python
print("Total points :",len(contextual_anomaly))
print("Total normal points :",sum(anomalies == False))
print("Total anomaly points :",sum(anomalies == True))
```
